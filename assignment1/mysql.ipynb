## Part 2

```python
# Create a table `course`
mycursor = mydb.cursor()
mycursor.execute("CREATE TABLE data1050f26.course (course_id VARCHAR(128) PRIMARY KEY, title VARCHAR(128), dept_name VARCHAR(128), credits INT)")
```

```python
# add FK `dept_name`` to `course`
query = ('ALTER TABLE data1050f26.course ADD foreign key(dept_name) references data1050f26.department(dept_name)')
mycursor.execute(query)
```

```python
# Create a table `prereq`
mycursor = mydb.cursor()
mycursor.execute("CREATE TABLE data1050f26.prereq (course_id VARCHAR(128), prereq_id VARCHAR(128), PRIMARY KEY (course_id, prereq_id))")
```

```python
# add FK `course_id` and `prereq_id` to `prereq`
query1 = ('ALTER TABLE data1050f26.prereq ADD foreign key(course_id) references data1050f26.course(course_id)')
mycursor.execute(query1)

query2 = ('ALTER TABLE data1050f26.prereq ADD foreign key(prereq_id) references data1050f26.course(course_id)')
mycursor.execute(query2)
```

```python
# Create a table `advisor`
mycursor = mydb.cursor()
mycursor.execute("CREATE TABLE data1050f26.advisor (s_id VARCHAR(128) PRIMARY KEY, i_id VARCHAR(128))")
```

```python
# add FK `s_id` and `i_id` to `advisor`
query1 = ('ALTER TABLE data1050f26.advisor ADD foreign key(s_id) references data1050f26.student(ID)')
mycursor.execute(query1)
query2 = ('ALTER TABLE data1050f26.advisor ADD foreign key(i_id) references data1050f26.instructor(ID)')
mycursor.execute(query2)
```

```python
# Create a table `classroom`
mycursor = mydb.cursor()
mycursor.execute("CREATE TABLE data1050f26.classroom (building VARCHAR(128), room_no VARCHAR(128), capacity INT, PRIMARY KEY (building, room_no))")
```

```python
# Non-NULL constraint on `title` and `dept_name` in `course`
# Credits could be NULL, because some courses may not have credits (e.g., research courses, workshops, etc.)
mycursor.execute("ALTER TABLE data1050f26.course MODIFY title VARCHAR(128) NOT NULL")
mycursor.execute("ALTER TABLE data1050f26.course MODIFY dept_name VARCHAR(128) NOT NULL")

# Non-NULL constraint on `i_id` in `advisor`
mycursor.execute("ALTER TABLE data1050f26.advisor MODIFY i_id VARCHAR(128) NOT NULL")

# Non-NULL constraint on `capacity` in `classroom`
mycursor.execute("ALTER TABLE data1050f26.classroom MODIFY capacity INT NOT NULL")
```

```python
#connect to a specific database
mydb = mysql.connector.connect(
  host="localhost",
  user="root",
  password="qwertyuiop", #REPLACE THIS WITH YOUR PASSWORD 
  database = "data1050f26" #connecting to testDatabase
)

mycursor = mydb.cursor()

mycursor.execute("SHOW TABLES")

for x in mycursor:
  print(x)
```

```python
# Show the columns (attributes, fields) of each new table
tables = ['course', 'prereq', 'advisor', 'classroom']

for table in tables:
    print(f"=== {table} ===")
    mycursor.execute(f"DESCRIBE {table}")
    for x in mycursor:
        print(x)
    print()
```
